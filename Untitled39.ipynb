```python
import os
import zipfile
import nltk
from nltk import sent_tokenize
from gensim.utils import simple_preprocess

# Download the 'punkt_tab' tokenizer data if not already present
try:
    nltk.data.find('tokenizers/punkt_tab') # Check for punkt_tab specifically
except LookupError:
    nltk.download('punkt_tab')

story=[]

zip_path = "/content/archive (10).zip"
extract_dir = "/content/extracted_archive"

# Create the extraction directory if it doesn't exist
if not os.path.exists(extract_dir):
    os.makedirs(extract_dir)

# Extract the zip file
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

# Now iterate through the extracted files
for filename in os.listdir(extract_dir):
  # Assuming the files inside are .txt files
  if filename.endswith(".txt"):
    filepath = os.path.join(extract_dir, filename)
    with open(filepath, 'r', encoding='utf-8', errors='ignore') as f:
      corpus=f.read()
      raw_sent=sent_tokenize(corpus)
      for sent in raw_sent:
        story.append(simple_preprocess(sent))
```

```python
len(story)
```

```python
model=gensim.models.Word2Vec(window=10,
                             min_count=2,
                             workers=4)
```

```python
model.build_vocab(story)
```

```python
model.train(story,total_examples=model.corpus_count,epochs=model.epochs)
```

```python
model.wv.most_similar("ygritte")
```
